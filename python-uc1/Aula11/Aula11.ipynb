```python
#fromkeys
chaves = ["nome", "dt_nascimento", "CEP", "logradouro", "bairro", "cidade", "estado"]
valor_padrao = "N/A"
usuario = dict.fromkeys(chaves, valor_padrao)
print(usuario)

usuario["nome"]=input("Digite o seu nome:")
print(usuario["nome"])
usuario["dt_nascimento"]= input("Digite a data de nascimento:")
print(usuario["dt_nascimento"])
usuario["CEP"]=input("Digite o seu CEP:")
print(usuario["CEP"])
usuario["logradouro"]=input("Digite o seu logradouro:")
print(usuario["logradouro"])
usuario["bairro"]=input("Digite o seu bairro:")
print(usuario["bairro"])
usuario["cidade"]= input("Digite o nome da sua cidade")
print(usuario["cidade"])
usuario["estado"]= input("Digite o nome do seu estado")
print(usuario["estado"])

for chave, valor in usuario.items():
    print(f"{chave.upper():_<20}: {valor.capitalize()}")
```
